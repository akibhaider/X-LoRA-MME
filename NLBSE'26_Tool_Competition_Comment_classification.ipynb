## Introduction

The [NLBSE'26 tool competition](https://nlbse2026.github.io/tools/) on **code comment classification** consists of building and testing **a set of multi-label classification models** for each programming language to classify code comment sentences as belonging to one or more categories representing the types of information that a sentence is conveying. The code comments correspond to comments in code classes.

For the competition, we provide a dataset of **9,361** code comment sentences belong to **18** categories (7 for Java, 5 for Python, and 6 for Pharo). We provide three baseline classifiers based on Sentence Transformers. Participants will propose their classifiers for this task to outperform the baseline according to the conditions defined below.

The dataset and details about data preparation and baselines can be found in [our repository](https://github.com/nlbse2026/code-comment-classification/tree/main) for the competition.

## Dataset Overview

We provide a dataset of **1,733** manually labeled class comments and **9,361** sentences from these comments, distributed into various categories specific to each programming language (*summary*, *intent*, *rationale*, etc.). These comments were extracted from **20** open-source projects written in three programming languages: Java, Pharo, or Python. We provide the associated code class (i.e., class name) for each sentence as well as the source code of the software projects.

Each comment sentence can belong to one or more categories of the language (from a minimum of **5** to a maximum of **7**, depending on the language). Each category represents the type of information that the sentence is conveying.

As one sentence can belong to multiple categories, the competition focuses on multi-label classification for each language, in total 487 instances belong to more than one category. In other words, participants will build multiple multi-class classifiers for each language, each  will focus on one language and determine if a sentence does or does not belong to these categories.

You will use this dataset for evaluating your classification models and compare the achieved results against the proposed baselines.

## Categories for comment sentences

Each programming language has its own taxonomy of manually validated categories that span from a minimum of five (5) to a maximum of seven (7):
 * Java has **6,595** comment sentences distributed in **7** categories: *summary*, *pointer*, *deprecation*, *rational*, *ownership*, *usage*, and *expand*;
 * Pharo has **1,108** comment sentences distributed in **6** categories: *key messages*, *intent*, *class references*, *example*, *key implementation*, *responsibilities*, and *collaborators*;
 * Python has **1,658** comment sentences distributed in **5** categories: *summary*, *parameters*, *usage*, *development notes*, and *expand*.

The definition of these categories can be found **[in this paper](https://www.sciencedirect.com/science/article/pii/S0164121221001448?via%3Dihub)**.

**The competition requires you to train and test a multi-class classifier for each language described above (i.e., 3 classifiers in total).**

## Training and test datasets

We provide a [HuggingFace Dataset](https://huggingface.co/datasets/NLBSE/nlbse26-code-comment-classification) with the competition data. The dataset has six splits in total, two splits (train and test) per language.

Each row represents a sentence (aka an instance), and each sentence contains five columns as follows:
- `class` is the class name referring to the source code file where the sentence comes from;
- `comment_sentence` is the actual sentence string, which is part of a (multi-line) class comment;
- `partition` is the dataset split in training and testing; `0` identifies training instances, and `1` identifies testing instances, respectively;
- `combo` is the class name appended to the sentence string, used to train the baselines;
- `labels` is the ground-truth category, it is a binary list that a single sample belongs to. Each sample belongs one or more categories.

The list of labels is defined for each of as follows:
- `java`: [`summary`, `Ownership`, `Expand`, `usage`, `Pointer`, `deprecation`, `rational`],
- `python`: [`Usage`, `Parameters`, `DevelopmentNotes`, `Expand`, `Summary`],
- `pharo`: [`Keyimplementationpoints`, `Example`, `Responsibilities`, `Intent`, `Keymessages`, `Collaborators`]


**NOTE:**  Another reason for this is that part of a sentence could convey a category (information type), so the experts only took that relevant part for the category. For this reason:
1. it is possible that some sentences in the dataset are incomplete; and
2. the sentences are preprocessed to some extent (see the [README](https://github.com/nlbse2026/code-comment-classification/tree/main) file of the datasets for more details).

It is the nature of classification, since one sentence may belong to multiple classes, so if one combines the entire data from categories, it may seem duplicated (i.e., the same sentence in multiple classes), a few of these sentences may appear duplicated (i.e., same text) in the training and test sets.




## Training your classifiers

**The unit of classification for your classifiers is a code comment sentence**.

Participants are free to select and transform the data from the provided **training dataset** under the following conditions:
1. No external information sources can be used to train/fine-tune the proposed classifiers. That means that **only the comment sentence text and associated source code (i.e., the class name)** can be used for classification. Participants can leverage this  code to improve classification.
    * **Important**: our dataset provides the code class name associated with each sentence and the original source code of the software projects; this way, participants can locate the class in the original source code and be able to extract features from it, if desired; in our [repository](https://github.com/nlbse2026/code-comment-classification/tree/main), see the `project_classes` folder of each language with the mapping between class names and projects.
2. While no new sources of information can be added/used, we allow using classification models pre-trained on other datasets but they must be fine-tuned on the provided training dataset.
3. Participants may preprocess, sample, apply over/under-sampling, select a subset of the attributes, perform feature engineering, split the training set into a model-fine-tuning validation set, etc.

Please contact us if you have any questions regarding these rules.

## Submission evaluation and ranking

The proposed classifiers will be evaluated based on their class-detection performance on the provided test sets for each of the language categories.

**Important:** you may apply any preprocessing or feature engineering on test dataset except sampling, rebalancing, undersampling, or oversampling techniques.

The classifiers should assign one single category label to the sentences in the **test set** for their corresponding category.
Classification performance must be measured by the $F_1$ score (defined below) achieved by each classifier (i.e., for each category in a language).

A submission (i.e., paper) in the tool competition must provide:
- Precision $P_c$ for each category $c$
- Recall $R_c$, for each category $c$
- $F_{1,c}$ score as the harmonic mean between $P_c$ and $R_c$, for each category $c$
- The average precision, recall, and $F_1$ scores across all 19 categories.

These metrics are defined as follows:
\begin{align}
P_c &= \frac{ TP_c }{ TP_c + FP_c } &
R_c &= \frac{ TP_c }{ TP_c + FN_c } &
F_{1,c} &= 2 \cdot \frac{ P_c \cdot R_c }{ P_c + R_c }
\end{align}

where $TP_c$, $FP_c$ and $FN_c$ represent the true positives, false positives, and false negatives for a category $c$, respectively.

The 19 scores of the proposed classifiers will be compared against the 19 $F_1$ scores achieved by the baseline classifier (see below) to rank the participants and determine a winner.

**To rank the competition submissions and determine a winner, we will use the following formula**:

\begin{align}
submission\_score(model) &= 0.60 \times avg. \space F_1
 + 0.2 \times max(\frac{(max\_avg\_runtime  - measured\_avg\_runtime)}{max\_avg\_runtime)}, 0)
 + 0.2 \times max(\frac{(max\_avg\_GFLOPS  - measured\_avg\_GFLOPS)}{max\_avg\_GFLOPS)}, 0)
 \end{align}


where:
- $avg. \space F_1$ is the average of the $F_1$ scores achieved by the proposed classifiers (implementing a single $model$, e.g., SVM, BERT, etc.) across the 18 categories.
- $max\_avg\_runtime$ indicates the maximum average inference runtime (using a Colab T4 instance -- read below) and the $measured\_avg\_runtime$ is the actual average inference runtime of the proposed classifiers. The runtime should be measured 10 times and should averaged across all categories and samples (i.e., code comments) -- **see the reference [notebook](https://colab.research.google.com/drive/1RULzasjO_nrqiXLrGze-PznFlHKtGQ4s?usp=sharing) for the specific algorithm to measure the runtime**. Solutions with a runtime higher than $max\_avg\_runtime$ are allowed.
- $max\_avg\_GFLOPS$ indicates the maximum average compute in Giga FLOating Point operations per Second. Solutions with coumputational effort higher than $max\_avg\_GFLOPS$ are allowed.
- If a part of the submisson_score is negative, e.g., due to a high number of GFLOPS, is set to zero, such that no negative submussion-scores are possible

With this formula we encourage the participants to maximize the overall classification effectiveness and minimize the runtime of their models, for practical purposes.

**To account for hardware differences, the final evaluation must be run on a Google Colab T4 instance.** We provide a [base infrastructure (i.e, a notebook)](https://colab.research.google.com/drive/1QlfXoEeDV2i1RR1yg04lZkLejDGMp9R6?usp=sharing) to run classifiers on the test set, measure their runtime, and compute the submission score. This infrastructure can be reused by the competition participants.

The submission with the highest score, determined by our formula, will be the winner of the competition.

**Note**: we only allow training/testing a single ML/DL model (e.g., SVM or BERT) considering **all the categories**, rather than considering single categories. In other words, we do not allow using distinct models for subsets of categories.



## Baseline classifiers

We trained and tested 3 classifiers (one for each language) using the Sentence Transformer architecture on the provided training and test sets.

The baseline classifiers, coined as STACC and proposed by [Al-Kaswan et al.](https://arxiv.org/abs/2302.13681), are lightweight classifiers developed using SetFit, an efficient
and prompt-free framework for few-shot fine-tuning of
Sentence Transformers. For the fine-tuning, we
relied on the Optuna backend with SetFit to find the best
hyperparameters using data from the Java deprecation
category. The performance of the model with the best
hyperparameters was obtained on the data from all the
categories in the test set. The input data was formatted by appending the code file name to the corresponding comments, separated by the ’|’ symbol.

More information about the model can be found in [this paper](https://arxiv.org/abs/2302.13681) and in our [notebook](https://colab.research.google.com/drive/1QlfXoEeDV2i1RR1yg04lZkLejDGMp9R6?usp=sharing) and [repository](https://github.com/nlbse2026/code-comment-classification/blob/main/SetFit_baseline.ipynb), which replicates the baseline results. The pre-trained models and dataset are avaliable on the [HuggingFace Hub](https://huggingface.co/collections/NLBSE/nlbse26-code-comment-classification-competition-68a6c1e0aa5cc08aea456a44).
