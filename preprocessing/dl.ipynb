# Download NLBSE'26 Code Comment Classification Dataset

Download the dataset using Hugging Face Datasets Library

```python
from datasets import load_dataset
from pathlib import Path

# Create dataset directory
DATA_DIR = Path('dataset')
DATA_DIR.mkdir(exist_ok=True, parents=True)

print("Downloading NLBSE'26 Code Comment Classification dataset...")
print("="*80)

# Download the full dataset
dataset = load_dataset("NLBSE/nlbse26-code-comment-classification")

print(f"\nDataset splits: {list(dataset.keys())}")
print(f"\nDataset info:")
for split in dataset.keys():
    print(f"  {split}: {len(dataset[split])} samples")
    print(f"    Features: {dataset[split].features}")
```

## Explore Dataset Structure

```python
import pandas as pd

# Get the actual split names
available_splits = list(dataset.keys())
print(f"Available splits: {available_splits}")
print("="*80)

# Use the first available split to show a sample
first_split = available_splits[0]
print(f"\nSample from '{first_split}' split:")
print("="*80)
print(dataset[first_split][0])

# Check unique languages/partitions for each split
for split_name in available_splits:
    df = pd.DataFrame(dataset[split_name])
    print(f"\n{split_name.upper()} set partitions (languages):")
    print(df['partition'].value_counts())
```

## Save to Parquet Files (Split by Language)

```python
import pandas as pd
from pathlib import Path

# Ensure DATA_DIR is defined
DATA_DIR = Path('dataset')

print("Saving dataset splits to parquet files...")
print("="*80)

# Get actual split names from the dataset
available_splits = list(dataset.keys())
print(f"Available splits: {available_splits}\n")

for split_name in available_splits:
    # Convert to DataFrame and save directly
    df = pd.DataFrame(dataset[split_name])
    
    # Save the entire split as one parquet file
    filename = f"{split_name}.parquet"
    filepath = DATA_DIR / filename
    
    df.to_parquet(filepath, index=False)
    print(f"- Saved {filename}: {len(df)} samples")
    
    # Show what languages/partitions are in this split
    if 'partition' in df.columns:
        print(f"  Languages: {df['partition'].value_counts().to_dict()}")

print("\n" + "="*80)
print("  Dataset downloaded and saved successfully!")
print(f"Location: {DATA_DIR.resolve()}")
print("="*80)
```
